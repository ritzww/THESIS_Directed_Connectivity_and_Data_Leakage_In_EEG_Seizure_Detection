```python
import h5py
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# 1. LOAD DATA
# -----------------------------
file_path = "../Features/chb01_03_features.mat"

with h5py.File(file_path, 'r') as f:
    X = f['all_feats'][:]          # (windows, 1530)
    y = f['all_labels'][:].squeeze()

print("X shape:", X.shape)
print("y distribution:", np.bincount(y.astype(int)))

# -----------------------------
# 2. STRUCTURE PARAMETERS
# -----------------------------
n_src = 18
n_dst = 17
n_bands = 5

band_size = n_src * n_dst  # 306

# -----------------------------
# 3. SPLIT ICTAL / INTERICTAL
# -----------------------------
ictal = X[y == 1]
interictal = X[y == 0]

print("Ictal windows:", ictal.shape[0])
print("Interictal windows:", interictal.shape[0])

# -----------------------------
# 4. MEAN FEATURE VECTORS
# -----------------------------
ictal_mean = np.mean(ictal, axis=0)
interictal_mean = np.mean(interictal, axis=0)

diff = ictal_mean - interictal_mean

# -----------------------------
# 5. HELPER FUNCTION: VECTOR → GC MATRIX
# -----------------------------
def vector_to_gc(vec_band):
    return vec_band.reshape(n_src, n_dst)

# -----------------------------
# 6. PLOT ICTAL VS INTERICTAL (PER BAND)
# -----------------------------
fig, axes = plt.subplots(3, n_bands, figsize=(18, 8))

for b in range(n_bands):

    start = b * band_size
    end = (b + 1) * band_size

    ictal_mat = vector_to_gc(ictal_mean[start:end])
    inter_mat = vector_to_gc(interictal_mean[start:end])
    diff_mat = vector_to_gc(diff[start:end])

    # Ictal
    im1 = axes[0, b].imshow(ictal_mat, cmap='viridis', aspect='auto')
    axes[0, b].set_title(f'Ictal Band {b}')

    # Interictal
    im2 = axes[1, b].imshow(inter_mat, cmap='viridis', aspect='auto')
    axes[1, b].set_title(f'Interictal Band {b}')

    # Difference
    im3 = axes[2, b].imshow(diff_mat, cmap='bwr', aspect='auto')
    axes[2, b].set_title(f'Diff Band {b}')



plt.suptitle("Ictal vs Interictal GC Connectivity (All Bands)", fontsize=14)

plt.subplots_adjust(
    right=0.88,   
    wspace=0.3,
    hspace=0.4
)

fig.colorbar(im1, ax=axes[0, :].ravel().tolist(), shrink=1, pad=0.05)
fig.colorbar(im2, ax=axes[1, :].ravel().tolist(), shrink=1, pad=0.05)
fig.colorbar(im3, ax=axes[2, :].ravel().tolist(), shrink=1, pad=0.05)

plt.show()
```

```python
import numpy as np
from scipy.stats import ranksums

# -----------------------------
# 1. STRUCTURE
# -----------------------------
n_src = 18
n_dst = 17
n_bands = 5
band_size = n_src * n_dst  # 306

# -----------------------------
# 2. SPLIT DATA
# -----------------------------
ictal = X[y == 1]
interictal = X[y == 0]

print("Ictal:", ictal.shape, "Interictal:", interictal.shape)

# -----------------------------
# 3. RESHAPE INTO:
# (windows, src, dst, bands)
# -----------------------------
def reshape_gc(data):
    n = data.shape[0]
    return data.reshape(n, n_bands, band_size).reshape(n, n_src, n_dst, n_bands)

gc_ictal_all = reshape_gc(ictal)
gc_inter_all = reshape_gc(interictal)

# -----------------------------
# 4. STAT TEST + EFFECT SIZE MAPS
# -----------------------------
diff_map = np.zeros((n_src, n_dst, n_bands))
sig_mask = np.zeros((n_src, n_dst, n_bands), dtype=bool)

for b in range(n_bands):
    for i in range(n_src):
        for j in range(n_dst):

            ictal_vals = gc_ictal_all[:, i, j, b]
            inter_vals = gc_inter_all[:, i, j, b]

            # avoid empty or constant arrays
            if len(np.unique(ictal_vals)) < 2 or len(np.unique(inter_vals)) < 2:
                continue

            stat, p = ranksums(ictal_vals, inter_vals)

            diff_map[i, j, b] = ictal_vals.mean() - inter_vals.mean()
            sig_mask[i, j, b] = p < 0.05

# -----------------------------
# 5. MASK NON-SIGNIFICANT RESULTS
# -----------------------------
diff_display = np.where(sig_mask, diff_map, 0)
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, n_bands, figsize=(25, 5))

for b in range(n_bands):

    im = axes[b].imshow(diff_display[:, :, b], cmap='bwr', aspect='auto')
    axes[b].set_title(f"Significant GC Diff Band {b}")

    axes[b].set_xlabel("Target (17)")
    axes[b].set_ylabel("Source (18)")

fig.colorbar(im, ax=axes.ravel().tolist(), shrink=0.7)
plt.show()
```

```python
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt

# -----------------------------
# PARAMETERS
# -----------------------------
n_src = 18
n_dst = 17
n_bands = 5
band_size = n_src * n_dst

band = 0  # choose band (0–4)

# -----------------------------
# LOAD CORRECT GC MATRIX
# -----------------------------
# ictal_mean must already exist:
# shape: (1530,) = (5 * 18 * 17)

gc_matrix = ictal_mean[band * band_size : (band + 1) * band_size]
gc_matrix = gc_matrix.reshape(n_src, n_dst)

# sanity check
print("GC shape:", gc_matrix.shape)
print("is square:", gc_matrix.shape[0] == gc_matrix.shape[1])

# -----------------------------
# THRESHOLD (top 15%)
# -----------------------------
thr = np.percentile(gc_matrix, 85)

# -----------------------------
# BUILD GRAPH
# -----------------------------
G = nx.DiGraph()

# nodes
for i in range(n_src):
    G.add_node(i)

# edges (ONLY valid i→j pairs)
for i in range(n_src):
    for j in range(n_dst):
        if gc_matrix[i, j] > thr:
            G.add_edge(i, j, weight=float(gc_matrix[i, j]))

# -----------------------------
# REMOVE ANY POSSIBLE SELF LOOPS (SAFEGUARD)
# -----------------------------
G.remove_edges_from(nx.selfloop_edges(G))

print("Self-loops remaining:", list(nx.selfloop_edges(G)))

# -----------------------------
# LAYOUT (FIXED ISSUE)
# -----------------------------
pos = nx.spring_layout(G, seed=42, k=0.8)

# -----------------------------
# EDGE WEIGHTS → WIDTHS
# -----------------------------
edges = G.edges(data=True)
weights = np.array([e[2]['weight'] for e in edges]) if len(edges) > 0 else np.array([])

if len(weights) > 0:
    w = (weights - weights.min()) / (weights.max() - weights.min() + 1e-8)
    widths = 1 + 4 * w
else:
    widths = []

# -----------------------------
# PLOT
# -----------------------------
plt.figure(figsize=(7, 7))

nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=600,
    node_color="lightblue",
    arrows=True,
    width=widths,
    edge_color="gray",
    alpha=0.8,
    font_size=10
)

plt.title(f"GC Network (Ictal, Band {band})")
plt.show()
```
