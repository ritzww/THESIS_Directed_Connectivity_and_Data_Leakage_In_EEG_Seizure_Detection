```python
import h5py
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# 1. LOAD DATA
# -----------------------------
file_path = "../Features/chb01_03_features.mat"

with h5py.File(file_path, 'r') as f:
    X = f['all_feats'][:]          # (windows, 1530)
    y = f['all_labels'][:].squeeze()

print("X shape:", X.shape)
print("y distribution:", np.bincount(y.astype(int)))

# -----------------------------
# 2. STRUCTURE PARAMETERS
# -----------------------------
n_src = 18
n_dst = 17
n_bands = 5

band_size = n_src * n_dst  # 306

# -----------------------------
# 3. SPLIT ICTAL / INTERICTAL
# -----------------------------
ictal = X[y == 1]
interictal = X[y == 0]

print("Ictal windows:", ictal.shape[0])
print("Interictal windows:", interictal.shape[0])

# -----------------------------
# 4. MEAN FEATURE VECTORS
# -----------------------------
ictal_mean = np.mean(ictal, axis=0)
interictal_mean = np.mean(interictal, axis=0)

diff = ictal_mean - interictal_mean

# -----------------------------
# 5. HELPER FUNCTION: VECTOR → GC MATRIX
# -----------------------------
def vector_to_gc(vec_band):
    return vec_band.reshape(n_src, n_dst)

# -----------------------------
# 6. PLOT ICTAL VS INTERICTAL (PER BAND)
# -----------------------------
fig, axes = plt.subplots(3, n_bands, figsize=(18, 8))

for b in range(n_bands):

    start = b * band_size
    end = (b + 1) * band_size

    ictal_mat = vector_to_gc(ictal_mean[start:end])
    inter_mat = vector_to_gc(interictal_mean[start:end])
    diff_mat = vector_to_gc(diff[start:end])

    # Ictal
    im1 = axes[0, b].imshow(ictal_mat, cmap='viridis', aspect='auto')
    axes[0, b].set_title(f'Ictal Band {b}')

    # Interictal
    im2 = axes[1, b].imshow(inter_mat, cmap='viridis', aspect='auto')
    axes[1, b].set_title(f'Interictal Band {b}')

    # Difference
    im3 = axes[2, b].imshow(diff_mat, cmap='bwr', aspect='auto')
    axes[2, b].set_title(f'Diff Band {b}')



plt.suptitle("Ictal vs Interictal GC Connectivity (All Bands)", fontsize=14)

plt.subplots_adjust(
    right=0.88,   
    wspace=0.3,
    hspace=0.4
)

fig.colorbar(im1, ax=axes[0, :].ravel().tolist(), shrink=1, pad=0.05)
fig.colorbar(im2, ax=axes[1, :].ravel().tolist(), shrink=1, pad=0.05)
fig.colorbar(im3, ax=axes[2, :].ravel().tolist(), shrink=1, pad=0.05)

plt.show()
```

```python
import h5py
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# 1. LOAD DATA
# -----------------------------
file_path = "../Features/chb01_26_features.mat"

with h5py.File(file_path, 'r') as f:
    X = f['all_feats'][:]
    y = f['all_labels'][:].squeeze()

# If MATLAB saved as (1530, windows), transpose
if X.shape[0] == 1530:
    X = X.T

print("X shape:", X.shape)
print("y distribution:", np.bincount(y.astype(int)))

# -----------------------------
# 2. PARAMETERS
# -----------------------------
n_channels = 18
n_bands = 5

band_size = n_channels * (n_channels - 1)  # 306

band_names = [
    "Delta (0.5–4 Hz)",
    "Theta (4–8 Hz)",
    "Alpha (8–13 Hz)",
    "Beta (13–30 Hz)",
    "Gamma (30–45 Hz)"
]

# -----------------------------
# 3. SPLIT ICTAL / INTERICTAL
# -----------------------------
ictal = X[y == 1]
interictal = X[y == 0]

print("Ictal windows:", ictal.shape[0])
print("Interictal windows:", interictal.shape[0])

# -----------------------------
# 4. MEAN FEATURE VECTORS
# -----------------------------
ictal_mean = np.mean(ictal, axis=0)
interictal_mean = np.mean(interictal, axis=0)

diff = ictal_mean - interictal_mean

# -----------------------------
# 5. RECONSTRUCT 18x18 GC MATRIX
# -----------------------------
def vector_to_gc(vec_band, n=18):
    """
    Reconstruct an n×n matrix from MATLAB:
        feat = mat(mask)
    where mask = ~eye(n)

    Preserves MATLAB column-major ordering.
    """
    mat = np.full((n, n), np.nan)

    mask = ~np.eye(n, dtype=bool)

    mat_flat = mat.flatten(order='F')
    mask_flat = mask.flatten(order='F')

    mat_flat[mask_flat] = vec_band

    mat = mat_flat.reshape((n, n), order='F')

    return mat

# -----------------------------
# 6. GLOBAL COLOR LIMITS
# -----------------------------
all_ictal = []
all_inter = []
all_diff = []

for b in range(n_bands):

    start = b * band_size
    end = (b + 1) * band_size

    all_ictal.append(vector_to_gc(ictal_mean[start:end]))
    all_inter.append(vector_to_gc(interictal_mean[start:end]))
    all_diff.append(vector_to_gc(diff[start:end]))

vmin_gc = np.nanmin([np.nanmin(m) for m in all_ictal + all_inter])
vmax_gc = np.nanmax([np.nanmax(m) for m in all_ictal + all_inter])

vlim_diff = max(
    abs(np.nanmin([np.nanmin(m) for m in all_diff])),
    abs(np.nanmax([np.nanmax(m) for m in all_diff]))
)

# -----------------------------
# 7. PLOT
# -----------------------------
fig, axes = plt.subplots(3, n_bands, figsize=(20, 10))

for b in range(n_bands):

    ictal_mat = all_ictal[b]
    inter_mat = all_inter[b]
    diff_mat = all_diff[b]

    # Ictal
    im1 = axes[0, b].imshow(
        ictal_mat,
        cmap='viridis',
        vmin=vmin_gc,
        vmax=vmax_gc
    )
    axes[0, b].set_title(band_names[b])

    # Interictal
    im2 = axes[1, b].imshow(
        inter_mat,
        cmap='viridis',
        vmin=vmin_gc,
        vmax=vmax_gc
    )

    # Difference
    im3 = axes[2, b].imshow(
        diff_mat,
        cmap='bwr',
        vmin=-vlim_diff,
        vmax=vlim_diff
    )

    for r in range(3):
        axes[r, b].set_xlabel("Target")
        axes[r, b].set_ylabel("Source")

# Row labels
axes[0, 0].set_ylabel("Source\n\nIctal")
axes[1, 0].set_ylabel("Source\n\nInterictal")
axes[2, 0].set_ylabel("Source\n\nDifference")

plt.suptitle(
    "Mean Spectral Granger Connectivity\nIctal vs Interictal",
    fontsize=16
)

# Shared colorbars
fig.colorbar(
    im1,
    ax=axes[0:2, :],
    fraction=0.02,
    pad=0.02,
    label="GC"
)

fig.colorbar(
    im3,
    ax=axes[2, :],
    fraction=0.02,
    pad=0.02,
    label="Ictal − Interictal"
)

plt.show()
```

```python
import numpy as np
from scipy.stats import ranksums

# -----------------------------
# 1. STRUCTURE
# -----------------------------
n_src = 18
n_dst = 17
n_bands = 5
band_size = n_src * n_dst  # 306

# -----------------------------
# 2. SPLIT DATA
# -----------------------------
ictal = X[y == 1]
interictal = X[y == 0]

print("Ictal:", ictal.shape, "Interictal:", interictal.shape)

# -----------------------------
# 3. RESHAPE INTO:
# (windows, src, dst, bands)
# -----------------------------
def reshape_gc(data):
    n = data.shape[0]
    return data.reshape(n, n_bands, band_size).reshape(n, n_src, n_dst, n_bands)

gc_ictal_all = reshape_gc(ictal)
gc_inter_all = reshape_gc(interictal)

# -----------------------------
# 4. STAT TEST + EFFECT SIZE MAPS
# -----------------------------
diff_map = np.zeros((n_src, n_dst, n_bands))
sig_mask = np.zeros((n_src, n_dst, n_bands), dtype=bool)

for b in range(n_bands):
    for i in range(n_src):
        for j in range(n_dst):

            ictal_vals = gc_ictal_all[:, i, j, b]
            inter_vals = gc_inter_all[:, i, j, b]

            # avoid empty or constant arrays
            if len(np.unique(ictal_vals)) < 2 or len(np.unique(inter_vals)) < 2:
                continue

            stat, p = ranksums(ictal_vals, inter_vals)

            diff_map[i, j, b] = ictal_vals.mean() - inter_vals.mean()
            sig_mask[i, j, b] = p < 0.05

# -----------------------------
# 5. MASK NON-SIGNIFICANT RESULTS
# -----------------------------
diff_display = np.where(sig_mask, diff_map, 0)
```

```python
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt

# -----------------------------
# PARAMETERS
# -----------------------------
n_channels = 18
n_bands = 5
band_size = n_channels * (n_channels - 1)

band_names = ["Delta", "Theta", "Alpha", "Beta", "Gamma"]

KEEP_TOP = 10  # % strongest edges kept

# -----------------------------
# RECONSTRUCTION (MATLAB-CORRECT)
# -----------------------------
def vector_to_gc(vec_band, n=18):
    mat = np.full((n, n), np.nan)
    mask = ~np.eye(n, dtype=bool)

    mat_flat = mat.flatten(order='F')
    mat_flat[mask.flatten(order='F')] = vec_band

    return mat_flat.reshape((n, n), order='F')

# -----------------------------
# GLOBAL SCALING (CRITICAL FIX)
# -----------------------------
all_vals = []

for arr in [ictal_mean, interictal_mean]:
    for b in range(n_bands):
        start = b * band_size
        end = (b + 1) * band_size
        all_vals.append(arr[start:end])

all_vals = np.concatenate(all_vals)
global_vals = all_vals[~np.isnan(all_vals)]

global_min = np.min(global_vals)
global_max = np.max(global_vals)

# -----------------------------
# GRAPH BUILDER (TOP % EDGES)
# -----------------------------
def build_graph(mat, keep_top=10):

    vals = mat[~np.isnan(mat)]
    thr = np.percentile(vals, 100 - keep_top)

    G = nx.DiGraph()

    for i in range(n_channels):
        G.add_node(i)

    for i in range(n_channels):
        for j in range(n_channels):

            if i == j:
                continue

            w = mat[i, j]

            if np.isnan(w):
                continue

            if w >= thr:
                G.add_edge(i, j, weight=w)

    return G

# -----------------------------
# FIXED LAYOUT (CONSISTENT ACROSS ALL PLOTS)
# -----------------------------
pos = nx.circular_layout(nx.complete_graph(n_channels))

# -----------------------------
# DRAW FUNCTION (GLOBAL SCALING)
# -----------------------------
def draw_graph(G, ax, title):

    edges = list(G.edges(data=True))

    if len(edges) == 0:
        ax.set_title(title + "\n(no edges)")
        ax.axis("off")
        return

    weights = np.array([d["weight"] for _, _, d in edges])

    # GLOBAL normalisation (KEY POINT)
    norm = (weights - global_min) / (global_max - global_min + 1e-12)

    # visual encoding
    widths = 0.5 + 5 * norm
    alphas = 0.05 + 0.95 * norm
    colors = [(0, 0, 0, a) for a in alphas]

    nx.draw_networkx_nodes(
        G,
        pos,
        node_size=300,
        node_color="lightblue",
        ax=ax
    )

    nx.draw_networkx_labels(
        G,
        pos,
        font_size=8,
        ax=ax
    )

    nx.draw_networkx_edges(
        G,
        pos,
        width=widths,
        edge_color=colors,
        arrows=True,
        arrowsize=12,
        ax=ax
    )

    ax.set_title(title)
    ax.axis("off")

# -----------------------------
# PLOT
# -----------------------------
fig, axes = plt.subplots(2, n_bands, figsize=(18, 8))

for b in range(n_bands):

    start = b * band_size
    end = (b + 1) * band_size

    # -------------------------
    # ICTAL
    # -------------------------
    ictal_mat = vector_to_gc(ictal_mean[start:end])
    G_ictal = build_graph(ictal_mat, KEEP_TOP)

    draw_graph(
        G_ictal,
        axes[0, b],
        f"Ictal\n{band_names[b]}"
    )

    # -------------------------
    # INTERICTAL
    # -------------------------
    inter_mat = vector_to_gc(interictal_mean[start:end])
    G_inter = build_graph(inter_mat, KEEP_TOP)

    draw_graph(
        G_inter,
        axes[1, b],
        f"Interictal\n{band_names[b]}"
    )

plt.suptitle(
    f"GC Networks (Top {KEEP_TOP}% edges, global scaling)",
    fontsize=16
)

plt.tight_layout()
plt.show()
```
