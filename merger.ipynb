# Normal Results

```python
folder = Path("../ABLATIAN RESULTS/MODEL 1")

# get all csv files
csvs = list(folder.glob("*.csv"))


# split by filename
result_files = [f for f in csvs if f.name.startswith("RESULTS_")]
segment_files = [f for f in csvs if f.name.startswith("SEGMENT_")]

print("RESULT files:", len(result_files))
print("SEGMENT files:", len(segment_files))
```

```python

# merge RESULTS
results_df = pd.concat(
    [pd.read_csv(f) for f in result_files],
    ignore_index=True
)

# merge SEGMENTS
segments_df = pd.concat(
    [pd.read_csv(f) for f in segment_files],
    ignore_index=True
)
```

# Merge all Results

```python
folder = Path("RESULTS")

# get all csv files
csvs = list(folder.glob("*.csv"))


# split by filename
result_files = [f for f in csvs if f.name.startswith("RESULTS_")]

print("RESULT files:", len(result_files))
```

```python
result_files
```

```python
results_0_10 = pd.read_csv(result_files[0]) 
results_0_5 = pd.read_csv(result_files[1]) 
results_10_10 = pd.read_csv(result_files[2]) 
results_10_5 = pd.read_csv(result_files[3]) 
```

```python
results_0_10['windows']=0
results_0_5['windows']=0
results_10_10['windows']=10000
results_10_5['windows']=10000
```

```python
results_10_5.columns
```

```python
a=results_0_10[['windows', 'percentile','config', 'C', 'gamma', 'patient', 'total_length_hours',
       'TP', 'FP', 'FN', 'sensitivity', 'precision', 'f1', 'FP/day', 'FP/hour',
       'latency_median', 'n_detected', 'mean_fold_roc_auc', 'mean_fold_ap']].copy()

b=results_0_5[['windows', 'percentile','config', 'C', 'gamma', 'patient', 'total_length_hours',
       'TP', 'FP', 'FN', 'sensitivity', 'precision', 'f1', 'FP/day', 'FP/hour',
       'latency_median', 'n_detected', 'mean_fold_roc_auc', 'mean_fold_ap']].copy()

c=results_10_10[['windows', 'percentile','config', 'C', 'gamma', 'patient', 'total_length_hours',
       'TP', 'FP', 'FN', 'sensitivity', 'precision', 'f1', 'FP/day', 'FP/hour',
       'latency_median', 'n_detected', 'mean_fold_roc_auc', 'mean_fold_ap']].copy()

d=results_10_5[['windows', 'percentile','config', 'C', 'gamma', 'patient', 'total_length_hours',
       'TP', 'FP', 'FN', 'sensitivity', 'precision', 'f1', 'FP/day', 'FP/hour',
       'latency_median', 'n_detected', 'mean_fold_roc_auc', 'mean_fold_ap']].copy()
```

```python
RESULTS = pd.concat([a,b,c,d], ignore_index=False)
```

```python
RESULTS.to_csv("EVENT_LEVEL_RESULTS.csv", index=False)
```
