```python
folder = Path("../RESULTADOS MODELO GOF FINAL")

# get all csv files
csvs = list(folder.glob("*.csv"))


# split by filename
result_files = [f for f in csvs if f.name.startswith("EVENTS_")]
segment_files = [f for f in csvs if f.name.startswith("FALSE_")]

print("RESULT files:", len(result_files))
print("SEGMENT files:", len(segment_files))
```

```python

# merge RESULTS
results_df = pd.concat(
    [pd.read_csv(f) for f in result_files],
    ignore_index=True
)

# merge SEGMENTS
segments_df = pd.concat(
    [pd.read_csv(f) for f in segment_files],
    ignore_index=True
)
```

```python
results_df
```

```python
segment_summ = (
    segments_df.groupby("patient")
      .agg(
          sensitivity=("seg_sensitivity", "mean"),
          precision=("seg_precision", "mean"),
          f1=("seg_f1", "mean"),
          tp=("tp", "mean"),
          fp=("fp", "mean"),
          fn=("fn", "mean"),
      )
      .sort_values("f1", ascending=False)
)

segment_summ
```

```python
# save
results_df.to_csv("EVENTS_MAIN_MODEL.csv", index=False)
segments_df.to_csv("FALSE_POSITIVES_MAIN_MODEL.csv", index=False)
```

```python
import pandas as pd

df = pd.read_csv("../RESULTS/BEFORE/0_SEGMENTS.csv")

summary = (
    df.groupby("patient")
      .agg(
          Records=("record", "count"),
          Total_Windows=("n_windows", "sum"),
          Seizure_Windows=("n_seizure_windows", "sum"),
      )
      .reset_index()
)

summary["Seizure_Windows (%)"] = (
    100 * summary["Seizure_Windows"] / summary["Total_Windows"]
)

summary
```

```python
results = pd.read_csv("../RESULTS/BEFORE/0_RESULTS.csv")

n_seizures = (
    results.groupby("patient")[["TP", "FN"]]
           .sum()
)

n_seizures["Seizures"] = n_seizures["TP"] + n_seizures["FN"]

summary = summary.merge(
    n_seizures[["Seizures"]],
    on="patient",
    how="left"
)

summary
```
