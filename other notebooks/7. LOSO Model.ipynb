```python
def load_h5_features(file_path):
    with h5py.File(file_path, "r") as f:
        X = f["all_feats"][:]

        y = f["all_labels"][:] if "all_labels" in f else None

    if y is not None:
        y = y.squeeze().astype(int)

    # -------------------------------------------------
    # AUTO FIX DIMENSION BY MATCHING LABEL LENGTH
    # -------------------------------------------------
    if y is not None:

        if X.shape[0] == len(y):
            # already correct: (windows, features)
            pass

        elif X.shape[1] == len(y):
            # transposed case: (features, windows)
            X = X.T

        else:
            raise ValueError(
                f"Cannot align X and y in {file_path}\n"
                f"X shape: {X.shape}, y length: {len(y)}"
            )

    else:
        # non-seizure file fallback
        if X.shape[0] < X.shape[1]:
            X = X.T

    return X, y
```

# Changing

```python
"""
Leave-one-record-out (LORO) seizure-detection pipeline — SzCORE-aligned scoring,
with a leakage-free DUAL-THRESHOLD detector.

Changes vs. the previous SzCORE version:
  A. BUGS FIXED
     - Feature-count grid and persistence grid are now SEPARATE names
       (KFEAT_GRID vs K_GRID). Previously both were called K_GRID, so the SVM
       was selecting 2/4/6 *features* instead of 20-200.
     - THRESHOLD_GRID spans NEGATIVE values again (seizures are caught at
       negative decision-function thresholds).
     - SVM feature count (best_kfeat) and persistence (best_k) are now distinct
       variables; they no longer overwrite each other.
  B. DUAL-THRESHOLD DETECTOR
     - th_low: calibrated per-fold on TRAIN out-of-fold scores to the FP/day
       budget (as before). Weak windows alarm only as part of a run of >= k.
     - th_high: a CONFIDENCE gate. A single window >= th_high alarms immediately
       (k=1 for strong evidence), shortening latency on confident onsets.
       th_high is set per-fold from TRAIN OOF scores as the lowest score whose
       training precision >= TH_HIGH_PRECISION. Never tuned on test.
  C. LATENCY now uses SzCORE's 30 s pre / 60 s post tolerance, so every SzCORE
     true positive also yields a latency (no more TP-with-nan-latency).
  D. k is still swept over K_GRID = [2, 4, 6].
"""

from dataclasses import dataclass
import re
from collections import defaultdict

import numpy as np
import pandas as pd

from sklearn.svm import SVC
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.model_selection import GroupKFold, cross_val_predict
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    average_precision_score,
    recall_score,
    precision_score,
    confusion_matrix,
)

from timescoring import scoring
from timescoring.annotations import Annotation

from joblib import parallel_backend

# =====================================================
# PARAMETERS
# =====================================================

WIN_SEC = 10.0
STEP_SEC = 2.5

# SVM feature-count grid (SelectKBest) — SEPARATE from persistence
KFEAT_GRID = [20, 50, 100]

C_GRID = [0.1]
GAMMA_GRID = ["scale"]

# Event-level operating point (chosen on TRAIN, applied to TEST)
TARGET_FP_PER_DAY = 1.9           # false-alarm budget (FP/day)
K_GRID = [2, 4, 6]                # persistence sweep, calibrated on TRAIN
THRESHOLD_GRID = np.linspace(-2.0, 2.0, 21)   # decision_function space (spans negatives)

# Dual-threshold confidence gate. th_high = lowest TRAIN-OOF score whose
# training precision >= this target. Fixed a priori, training-derived.
TH_HIGH_PRECISION = 0.70

# SzCORE event tolerances (used for latency matching to stay consistent with scoring)
TOL_PRE = 30.0
TOL_POST = 60.0

SZCORE_FS = 1                     # SzCORE label resolution (1 Hz)

N_SPLITS = 5                      # capped per training-set size below

ann = pd.read_csv("seizure_annotations.csv")

# =====================================================
# STRUCTURE
# =====================================================

@dataclass
class RecordPrediction:
    record_id: str
    window_starts: np.ndarray
    window_len: float
    scores: np.ndarray
    duration_sec: float
    true_seizures: list

# =====================================================
# HELPERS
# =====================================================

def _runs_of_true(mask):
    runs = []
    i = 0
    while i < len(mask):
        if mask[i]:
            j = i
            while j + 1 < len(mask) and mask[j + 1]:
                j += 1
            runs.append((i, j))
            i = j + 1
        else:
            i += 1
    return runs


def safe_auc(y_true, scores):
    return roc_auc_score(y_true, scores) if len(np.unique(y_true)) > 1 else np.nan


def safe_auprc(y_true, scores):
    return average_precision_score(y_true, scores) if len(np.unique(y_true)) > 1 else np.nan


def patient_of(record_id):
    """chb01_03 -> chb01 ; used for SzCORE per-subject aggregation."""
    m = re.match(r"(chb\d+)", record_id)
    return m.group(1) if m else record_id


def build_record_prediction(record_id, scores):
    n = len(scores)
    window_starts = np.arange(n) * STEP_SEC
    duration_sec = (window_starts[-1] + WIN_SEC) if n else 0.0
    return RecordPrediction(
        record_id=record_id,
        window_starts=window_starts,
        window_len=WIN_SEC,
        scores=np.asarray(scores),
        duration_sec=duration_sec,
        true_seizures=get_seizures_for_record(ann, record_id),
    )

# =====================================================
# DUAL-THRESHOLD DETECTOR -> 1 Hz ALARM MASK
# =====================================================

def _alarm_runs(rec, th_low, th_high, k_consecutive):
    """Return time intervals (start_sec, decision_sec) for each alarm under the
    dual-threshold rule. A run of windows >= th_low fires if it is at least
    k_consecutive long OR contains at least one window >= th_high (confident).
    decision_sec = end of the window at which the alarm condition is first met."""
    weak = rec.scores >= th_low
    strong = rec.scores >= th_high
    alarms = []
    for s, e in _runs_of_true(weak):
        run_len = e - s + 1
        strong_idx = np.where(strong[s:e + 1])[0]
        has_strong = strong_idx.size > 0

        if not (run_len >= k_consecutive or has_strong):
            continue

        # Decide WHEN the alarm fires (earliest condition met) -> drives latency.
        # Persistence path: alarm at the k-th window of the run.
        persist_decision_idx = s + (k_consecutive - 1) if run_len >= k_consecutive else None
        # Confidence path: alarm at the first strong window in the run.
        strong_decision_idx = s + strong_idx[0] if has_strong else None

        candidates = [c for c in (persist_decision_idx, strong_decision_idx) if c is not None]
        decision_idx = min(candidates)   # whichever triggers first

        start_sec = rec.window_starts[s]
        decision_sec = rec.window_starts[decision_idx] + rec.window_len
        # The alarm event spans from run start to run end (for overlap scoring)
        end_sec = rec.window_starts[e] + rec.window_len
        alarms.append((start_sec, decision_sec, end_sec))
    return alarms


def alarm_mask(rec, th_low, th_high, k_consecutive, fs=SZCORE_FS):
    n = int(np.ceil(rec.duration_sec)) * fs
    mask = np.zeros(n, dtype=int)
    for start_sec, _decision_sec, end_sec in _alarm_runs(rec, th_low, th_high, k_consecutive):
        a = int(round(start_sec * fs))
        b = int(round(end_sec * fs))
        mask[a:min(b, n)] = 1
    return mask


def reference_mask(rec, fs=SZCORE_FS):
    n = int(np.ceil(rec.duration_sec)) * fs
    mask = np.zeros(n, dtype=int)
    for sz_start, sz_end in rec.true_seizures:
        mask[int(round(sz_start * fs)):min(int(round(sz_end * fs)), n)] = 1
    return mask

# =====================================================
# SECONDARY: detection latency (NOT part of SzCORE)
# =====================================================

def first_alarm_latency(rec, th_low, th_high, k_consecutive):
    """Median latency over detected seizures. An alarm matches a seizure under
    SzCORE tolerance (30 s pre / 60 s post). Latency = alarm decision time minus
    true seizure onset, clamped at 0, from the FIRST matching alarm."""
    alarms = sorted(_alarm_runs(rec, th_low, th_high, k_consecutive))  # by start_sec
    latencies = []
    for sz_start, sz_end in rec.true_seizures:
        lo, hi = sz_start - TOL_PRE, sz_end + TOL_POST
        best = None
        for start_sec, decision_sec, end_sec in alarms:
            if start_sec < hi and end_sec > lo:          # overlap within tolerance
                best = max(0.0, decision_sec - sz_start)
                break                                    # FIRST matching alarm only
        if best is not None:
            latencies.append(best)
    return np.median(latencies) if latencies else np.nan

# =====================================================
# EVENT SCORING (SzCORE via timescoring)
# =====================================================

def score_one_file(rec, th_low, th_high, k_consecutive, fs=SZCORE_FS):
    ref = Annotation(reference_mask(rec, fs), fs)
    hyp = Annotation(alarm_mask(rec, th_low, th_high, k_consecutive, fs), fs)
    s = scoring.EventScoring(ref, hyp)   # default Parameters == SzCORE defaults
    return {
        "record_id": rec.record_id,
        "patient": patient_of(rec.record_id),
        "tp": s.tp,
        "fp": s.fp,
        "n_ref": s.refTrue,
        "duration_hours": rec.duration_sec / 3600,
        "sensitivity": s.sensitivity,
        "precision": s.precision,
        "f1": s.f1,
        "fp_per_day": s.fpRate,
        "median_latency": first_alarm_latency(rec, th_low, th_high, k_consecutive),
    }

# =====================================================
# th_high: CONFIDENCE GATE FROM TRAIN OOF SCORES
# =====================================================

def choose_th_high(oof_scores, y_train, precision_target=TH_HIGH_PRECISION,
                   candidate_thresholds=THRESHOLD_GRID):
    """Lowest threshold whose training precision >= precision_target.
    Computed on TRAIN out-of-fold scores only. If no threshold reaches the
    target, returns +inf (confidence gate effectively disabled)."""
    y = np.asarray(y_train)
    for th in np.sort(candidate_thresholds):            # low -> high
        pred = oof_scores >= th
        tp = np.sum(pred & (y == 1))
        fp = np.sum(pred & (y == 0))
        if (tp + fp) == 0:
            continue
        prec = tp / (tp + fp)
        if prec >= precision_target:
            return float(th)
    return np.inf

# =====================================================
# JOINT (k, th_low) CALIBRATION ON TRAINING RECORDS
# =====================================================

def calibrate_k_and_threshold(train_recs, th_high, target_fp_per_day, k_grid,
                              candidate_thresholds=THRESHOLD_GRID):
    """For each k, lowest th_low whose POOLED train FP/day <= target. Among the
    qualifying (k, th_low), pick the highest pooled TRAIN event sensitivity.
    th_high is held fixed (already chosen from train precision)."""
    best = None  # (sensitivity, k, th_low)
    for k in k_grid:
        for th in np.sort(candidate_thresholds):        # low -> high
            tp = fp = n_ref = 0
            total_hours = 0.0
            for rec in train_recs:
                o = score_one_file(rec, th_low=th, th_high=th_high, k_consecutive=k)
                tp += o["tp"]
                fp += o["fp"]
                n_ref += o["n_ref"]
                total_hours += o["duration_hours"]
            fp_per_day = fp / (total_hours / 24) if total_hours else np.inf
            if fp_per_day <= target_fp_per_day:
                sens = tp / n_ref if n_ref else 0.0
                if best is None or sens > best[0]:
                    best = (sens, k, th)
                break
    if best is None:
        return k_grid[0], float(np.max(candidate_thresholds))
    return best[1], best[2]

# =====================================================
# SVM TUNING (record-level CV, leak-free pipeline)
# =====================================================

def make_pipeline(C, gamma, k_features):
    return Pipeline([
        ("select", SelectKBest(f_classif, k=k_features)),
        ("scale", StandardScaler()),
        ("svc", SVC(C=C, gamma=gamma, kernel="rbf",
                    class_weight="balanced", cache_size=4000)),
    ])


def tune_svm_record_level(X, y, groups, n_splits):
    best_auc = -np.inf
    best_params = (C_GRID[0], GAMMA_GRID[0], KFEAT_GRID[0])
    gkf = GroupKFold(n_splits=n_splits)

    for k_features in KFEAT_GRID:
        for C in C_GRID:
            for gamma in GAMMA_GRID:
                fold_aucs = []
                for tr_idx, val_idx in gkf.split(X, y, groups):
                    pipe = make_pipeline(C, gamma, k_features)
                    pipe.fit(X[tr_idx], y[tr_idx])
                    sc = pipe.decision_function(X[val_idx])
                    fold_aucs.append(safe_auc(y[val_idx], sc))
                mean_auc = np.nanmean(fold_aucs)
                if mean_auc > best_auc:
                    best_auc = mean_auc
                    best_params = (C, gamma, k_features)
    return best_params

# =====================================================
# LORO PIPELINE
# =====================================================

results = []
event_results = []

for test_record in records:

    print(f"\nTesting: {test_record}")

    X_test, y_test = records[test_record]
    train_records = [r for r in records if r != test_record]

    X_train_all = np.vstack([records[r][0] for r in train_records])
    y_train_all = np.concatenate([records[r][1] for r in train_records])

    lengths = [len(records[r][1]) for r in train_records]
    bounds = np.cumsum([0] + lengths)

    groups = np.concatenate([
        np.full(lengths[i], i) for i in range(len(train_records))
    ])

    n_splits = max(2, min(N_SPLITS, len(train_records)))
    gkf = GroupKFold(n_splits=n_splits)

    # ---------------- SVM tuning (leak-free) ----------------
    best_C, best_gamma, best_kfeat = tune_svm_record_level(
        X_train_all, y_train_all, groups, n_splits
    )
    print(f"Best SVM: C={best_C}, gamma={best_gamma}, k_features={best_kfeat}")

    pipe = make_pipeline(best_C, best_gamma, best_kfeat)

    # ---------------- OUT-OF-FOLD train scores ----------------
    with parallel_backend("loky", n_jobs=8):
        oof_scores = cross_val_predict(
            pipe, X_train_all, y_train_all,
            groups=groups, cv=gkf, method="decision_function", n_jobs=8,
        )

    # th_high from TRAIN OOF precision target (confidence gate)
    th_high = choose_th_high(oof_scores, y_train_all)

    train_recs = [
        build_record_prediction(r, oof_scores[bounds[i]:bounds[i + 1]])
        for i, r in enumerate(train_records)
    ]

    # th_low + k calibrated to FP/day budget on TRAIN
    best_k, best_th_low = calibrate_k_and_threshold(
        train_recs, th_high=th_high,
        target_fp_per_day=TARGET_FP_PER_DAY, k_grid=K_GRID
    )
    th_high_disp = "inf" if not np.isfinite(th_high) else f"{th_high:.3f}"
    print(f"Calibrated (train): k={best_k}, th_low={best_th_low:.3f}, th_high={th_high_disp}")

    # ---------------- final model -> test record ----------------
    pipe.fit(X_train_all, y_train_all)
    scores = pipe.decision_function(X_test)
    y_pred = (scores >= 0).astype(int)

    # ---------------- segment metrics (guarded, unchanged) ----------------
    acc = accuracy_score(y_test, y_pred)
    auc = safe_auc(y_test, scores)
    auprc = safe_auprc(y_test, scores)
    precision = precision_score(y_test, y_pred, zero_division=0)
    sensitivity = recall_score(y_test, y_pred, zero_division=0)

    tn, fp, fn, tp = confusion_matrix(y_test, y_pred, labels=[0, 1]).ravel()
    specificity = tn / (tn + fp) if (tn + fp) > 0 else np.nan

    print(
        f"ACC={acc:.3f} AUC={auc:.3f} AUPRC={auprc:.3f} "
        f"Sens={sensitivity:.3f} Spec={specificity:.3f} Prec={precision:.3f}"
    )

    results.append({
        "acc": acc, "auc": auc, "auprc": auprc,
        "sensitivity": sensitivity, "specificity": specificity, "precision": precision,
    })

    # ---------------- event metrics (SzCORE, frozen train params) ----------------
    rec = build_record_prediction(test_record, scores)
    out = score_one_file(rec, th_low=best_th_low, th_high=th_high, k_consecutive=best_k)
    event_results.append(out)

    print("Seizures (ref):", out["n_ref"])
    print("TP / FP       :", out["tp"], "/", out["fp"])
    print("Sensitivity   :", out["sensitivity"])
    print("FP/day        :", out["fp_per_day"])
    print("Median latency:", out["median_latency"])

# =====================================================
# FINAL SUMMARY
# =====================================================

print("\n================ SEGMENT SUMMARY ================")
print("Mean ACC  :", np.nanmean([r["acc"] for r in results]))
print("Mean AUC  :", np.nanmean([r["auc"] for r in results]))
print("Mean AUPRC:", np.nanmean([r["auprc"] for r in results]))
print("Mean Sens :", np.nanmean([r["sensitivity"] for r in results]))
print("Mean Spec :", np.nanmean([r["specificity"] for r in results]))
print("Mean Prec :", np.nanmean([r["precision"] for r in results]))

print("\n================ EVENT SUMMARY (SzCORE, per-subject mean) ================")
by_pt = defaultdict(lambda: {"tp": 0, "fp": 0, "n_ref": 0, "hours": 0.0})
for r in event_results:
    p = by_pt[r["patient"]]
    p["tp"] += r["tp"]
    p["fp"] += r["fp"]
    p["n_ref"] += r["n_ref"]
    p["hours"] += r["duration_hours"]

sens_list, prec_list, f1_list, fppd_list = [], [], [], []
for pid, p in sorted(by_pt.items()):
    sens = p["tp"] / p["n_ref"] if p["n_ref"] else np.nan
    prec = p["tp"] / (p["tp"] + p["fp"]) if (p["tp"] + p["fp"]) else np.nan
    f1 = (2 * prec * sens / (prec + sens)
          if (prec is not np.nan and sens is not np.nan and (prec + sens)) else np.nan)
    fppd = p["fp"] / (p["hours"] / 24) if p["hours"] else np.nan
    sens_list.append(sens); prec_list.append(prec); f1_list.append(f1); fppd_list.append(fppd)
    print(f"  {pid}: sens={sens:.3f} prec={prec:.3f} f1={f1:.3f} fp/day={fppd:.3f}")

print("Mean Sensitivity:", np.nanmean(sens_list))
print("Mean Precision  :", np.nanmean(prec_list))
print("Mean F1         :", np.nanmean(f1_list))
print("Mean FP/day     :", np.nanmean(fppd_list))
print("Median latency  :", np.nanmedian([r["median_latency"] for r in event_results]))
```

```python
"""
Leave-one-record-out (LORO) seizure-detection pipeline — SzCORE-aligned scoring,
leakage-free DUAL-THRESHOLD detector — SPEED-OPTIMIZED.

Method is UNCHANGED. Only the execution is faster:
  S1. OUTER LORO LOOP IS PARALLELIZED across test records (joblib). Each fold
      is independent. Inner SVM/BLAS threads are pinned to 1 to avoid
      oversubscription; outer parallelism does the work.
  S2. k_features (SelectKBest) IS CHOSEN ONCE on the full pool, not swept per
      fold. The feature ranking is stable across the near-identical LORO
      training pools, so per-fold re-selection only added ~KFEAT x n_splits
      redundant SVM fits. Selection is still done on training-style data
      (GroupKFold AUC), never on any single held-out test record.
  S3. NEGATIVE SUBSAMPLING in the training fold: interictal windows are
      randomly subsampled to NEG_PER_POS x the positive count, with a fixed
      per-fold seed (deterministic, reproducible). class_weight='balanced'
      is retained. This shrinks the RBF kernel from O(n^2) on a ~99%-negative
      matrix. Positives are never dropped; the TEST record is never subsampled.

Scoring (timescoring/SzCORE), the dual-threshold detector, th_high precision
gate, th_low+k calibration, latency, and per-subject aggregation are byte-for-
byte identical to the previous version.

NOTE on the leakage-comparison experiment: subsampling changes the training set,
so the SAME NEG_PER_POS and SEED_BASE must be used across all three conditions,
or the comparison gains a confound.
"""

import os
# Pin math-library threads BEFORE numpy import so outer joblib parallelism
# doesn't oversubscribe cores (inner work stays single-threaded).
os.environ.setdefault("OMP_NUM_THREADS", "1")
os.environ.setdefault("OPENBLAS_NUM_THREADS", "1")
os.environ.setdefault("MKL_NUM_THREADS", "1")
os.environ.setdefault("NUMEXPR_NUM_THREADS", "1")

from dataclasses import dataclass
import re
from collections import defaultdict

import numpy as np
import pandas as pd

from sklearn.svm import SVC
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.model_selection import GroupKFold, cross_val_predict
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    average_precision_score,
    recall_score,
    precision_score,
    confusion_matrix,
)

from timescoring import scoring
from timescoring.annotations import Annotation

from joblib import Parallel, delayed

# =====================================================
# PARAMETERS
# =====================================================

WIN_SEC = 10.0
STEP_SEC = 2.5

KFEAT_GRID = [20, 50, 100]   # SelectKBest sweep (chosen ONCE, see S2)

C_GRID = [0.1]
GAMMA_GRID = ["scale"]

# Event-level operating point (chosen on TRAIN, applied to TEST)
TARGET_FP_PER_DAY = 1.0
K_GRID = [2, 4, 6]
THRESHOLD_GRID = np.linspace(-2.0, 2.0, 21)

TH_HIGH_PRECISION = 0.8

TOL_PRE = 30.0
TOL_POST = 60.0

SZCORE_FS = 1
N_SPLITS = 5

# Speed knobs
N_OUTER_JOBS = 8                  # parallel LORO folds (set to your core count)
NEG_PER_POS = 8                   # interictal windows kept per positive window
SEED_BASE = 12345                 # base RNG seed; per-fold seed = SEED_BASE + fold_idx

ann = pd.read_csv("seizure_annotations.csv")

# =====================================================
# STRUCTURE
# =====================================================

@dataclass
class RecordPrediction:
    record_id: str
    window_starts: np.ndarray
    window_len: float
    scores: np.ndarray
    duration_sec: float
    true_seizures: list

# =====================================================
# HELPERS
# =====================================================

def _runs_of_true(mask):
    runs = []
    i = 0
    while i < len(mask):
        if mask[i]:
            j = i
            while j + 1 < len(mask) and mask[j + 1]:
                j += 1
            runs.append((i, j))
            i = j + 1
        else:
            i += 1
    return runs


def safe_auc(y_true, scores):
    return roc_auc_score(y_true, scores) if len(np.unique(y_true)) > 1 else np.nan


def safe_auprc(y_true, scores):
    return average_precision_score(y_true, scores) if len(np.unique(y_true)) > 1 else np.nan


def patient_of(record_id):
    m = re.match(r"(chb\d+)", record_id)
    return m.group(1) if m else record_id


def build_record_prediction(record_id, scores):
    n = len(scores)
    window_starts = np.arange(n) * STEP_SEC
    duration_sec = (window_starts[-1] + WIN_SEC) if n else 0.0
    return RecordPrediction(
        record_id=record_id,
        window_starts=window_starts,
        window_len=WIN_SEC,
        scores=np.asarray(scores),
        duration_sec=duration_sec,
        true_seizures=get_seizures_for_record(ann, record_id),
    )

# =====================================================
# DUAL-THRESHOLD DETECTOR -> 1 Hz ALARM MASK  (unchanged)
# =====================================================

def _alarm_runs(rec, th_low, th_high, k_consecutive):
    weak = rec.scores >= th_low
    strong = rec.scores >= th_high
    alarms = []
    for s, e in _runs_of_true(weak):
        run_len = e - s + 1
        strong_idx = np.where(strong[s:e + 1])[0]
        has_strong = strong_idx.size > 0
        if not (run_len >= k_consecutive or has_strong):
            continue
        persist_decision_idx = s + (k_consecutive - 1) if run_len >= k_consecutive else None
        strong_decision_idx = s + strong_idx[0] if has_strong else None
        candidates = [c for c in (persist_decision_idx, strong_decision_idx) if c is not None]
        decision_idx = min(candidates)
        start_sec = rec.window_starts[s]
        decision_sec = rec.window_starts[decision_idx] + rec.window_len
        end_sec = rec.window_starts[e] + rec.window_len
        alarms.append((start_sec, decision_sec, end_sec))
    return alarms


def alarm_mask(rec, th_low, th_high, k_consecutive, fs=SZCORE_FS):
    n = int(np.ceil(rec.duration_sec)) * fs
    mask = np.zeros(n, dtype=int)
    for start_sec, _decision_sec, end_sec in _alarm_runs(rec, th_low, th_high, k_consecutive):
        a = int(round(start_sec * fs))
        b = int(round(end_sec * fs))
        mask[a:min(b, n)] = 1
    return mask


def reference_mask(rec, fs=SZCORE_FS):
    n = int(np.ceil(rec.duration_sec)) * fs
    mask = np.zeros(n, dtype=int)
    for sz_start, sz_end in rec.true_seizures:
        mask[int(round(sz_start * fs)):min(int(round(sz_end * fs)), n)] = 1
    return mask

# =====================================================
# SECONDARY: detection latency  (unchanged)
# =====================================================

def first_alarm_latency(rec, th_low, th_high, k_consecutive):
    alarms = sorted(_alarm_runs(rec, th_low, th_high, k_consecutive))
    latencies = []
    for sz_start, sz_end in rec.true_seizures:
        lo, hi = sz_start - TOL_PRE, sz_end + TOL_POST
        best = None
        for start_sec, decision_sec, end_sec in alarms:
            if start_sec < hi and end_sec > lo:
                best = max(0.0, decision_sec - sz_start)
                break
        if best is not None:
            latencies.append(best)
    return np.median(latencies) if latencies else np.nan

# =====================================================
# EVENT SCORING (SzCORE via timescoring)  (unchanged)
# =====================================================

def score_one_file(rec, th_low, th_high, k_consecutive, fs=SZCORE_FS):
    ref = Annotation(reference_mask(rec, fs), fs)
    hyp = Annotation(alarm_mask(rec, th_low, th_high, k_consecutive, fs), fs)
    s = scoring.EventScoring(ref, hyp)
    return {
        "record_id": rec.record_id,
        "patient": patient_of(rec.record_id),
        "tp": s.tp,
        "fp": s.fp,
        "n_ref": s.refTrue,
        "duration_hours": rec.duration_sec / 3600,
        "sensitivity": s.sensitivity,
        "precision": s.precision,
        "f1": s.f1,
        "fp_per_day": s.fpRate,
        "median_latency": first_alarm_latency(rec, th_low, th_high, k_consecutive),
    }

# =====================================================
# th_high precision gate  (unchanged)
# =====================================================

def choose_th_high(oof_scores, y_train, precision_target=TH_HIGH_PRECISION,
                   candidate_thresholds=THRESHOLD_GRID):
    y = np.asarray(y_train)
    for th in np.sort(candidate_thresholds):
        pred = oof_scores >= th
        tp = np.sum(pred & (y == 1))
        fp = np.sum(pred & (y == 0))
        if (tp + fp) == 0:
            continue
        prec = tp / (tp + fp)
        if prec >= precision_target:
            return float(th)
    return np.inf

# =====================================================
# (k, th_low) calibration  (unchanged)
# =====================================================

def calibrate_k_and_threshold(train_recs, th_high, target_fp_per_day, k_grid,
                              candidate_thresholds=THRESHOLD_GRID):
    best = None
    for k in k_grid:
        for th in np.sort(candidate_thresholds):
            tp = fp = n_ref = 0
            total_hours = 0.0
            for rec in train_recs:
                o = score_one_file(rec, th_low=th, th_high=th_high, k_consecutive=k)
                tp += o["tp"]
                fp += o["fp"]
                n_ref += o["n_ref"]
                total_hours += o["duration_hours"]
            fp_per_day = fp / (total_hours / 24) if total_hours else np.inf
            if fp_per_day <= target_fp_per_day:
                sens = tp / n_ref if n_ref else 0.0
                if best is None or sens > best[0]:
                    best = (sens, k, th)
                break
    if best is None:
        return k_grid[0], float(np.max(candidate_thresholds))
    return best[1], best[2]

# =====================================================
# PIPELINE + NEGATIVE SUBSAMPLING
# =====================================================

def make_pipeline(C, gamma, k_features):
    return Pipeline([
        ("select", SelectKBest(f_classif, k=k_features)),
        ("scale", StandardScaler()),
        ("svc", SVC(C=C, gamma=gamma, kernel="rbf",
                    class_weight="balanced", cache_size=4000)),
    ])


def subsample_negatives(X, y, neg_per_pos, seed):
    """Keep all positives; randomly keep neg_per_pos * n_pos negatives.
    Deterministic given seed. If there aren't that many negatives, keep all."""
    y = np.asarray(y)
    pos_idx = np.where(y == 1)[0]
    neg_idx = np.where(y == 0)[0]
    # If a fold has no positives, subsampling would empty the negatives too.
    # Keep the data unchanged in that degenerate case (the fold is single-class
    # anyway and contributes nan to AUC via the safe_auc guard).
    if len(pos_idx) == 0:
        return X, y
    n_keep = min(len(neg_idx), neg_per_pos * len(pos_idx))
    rng = np.random.default_rng(seed)
    keep_neg = rng.choice(neg_idx, size=n_keep, replace=False) if n_keep else np.array([], int)
    keep = np.sort(np.concatenate([pos_idx, keep_neg]))
    return X[keep], y[keep]

# =====================================================
# k_features CHOSEN ONCE (S2)
# =====================================================

def choose_k_features_once(records_dict):
    """Pick k_features on the full pooled dataset via GroupKFold AUC, once,
    before the LORO loop. Uses every record as its own group; this is a
    training-style selection (no single record is treated as the held-out
    test here)."""
    rec_ids = list(records_dict)
    X = np.vstack([records_dict[r][0] for r in rec_ids])
    y = np.concatenate([records_dict[r][1] for r in rec_ids])
    groups = np.concatenate([
        np.full(len(records_dict[r][1]), i) for i, r in enumerate(rec_ids)
    ])
    n_splits = max(2, min(N_SPLITS, len(rec_ids)))
    gkf = GroupKFold(n_splits=n_splits)

    best_auc, best_kfeat = -np.inf, KFEAT_GRID[0]
    for k_features in KFEAT_GRID:
        fold_aucs = []
        for tr_idx, val_idx in gkf.split(X, y, groups):
            Xs, ys = subsample_negatives(X[tr_idx], y[tr_idx], NEG_PER_POS, SEED_BASE)
            pipe = make_pipeline(C_GRID[0], GAMMA_GRID[0], k_features)
            pipe.fit(Xs, ys)
            sc = pipe.decision_function(X[val_idx])
            fold_aucs.append(safe_auc(y[val_idx], sc))
        mean_auc = np.nanmean(fold_aucs)
        if mean_auc > best_auc:
            best_auc, best_kfeat = mean_auc, k_features
    return best_kfeat

# =====================================================
# ONE LORO FOLD (runs in a worker)
# =====================================================

def run_one_fold(fold_idx, test_record, records_dict, k_features):
    X_test, y_test = records_dict[test_record]
    train_records = [r for r in records_dict if r != test_record]

    X_train_all = np.vstack([records_dict[r][0] for r in train_records])
    y_train_all = np.concatenate([records_dict[r][1] for r in train_records])

    lengths = [len(records_dict[r][1]) for r in train_records]
    bounds = np.cumsum([0] + lengths)
    groups = np.concatenate([np.full(lengths[i], i) for i in range(len(train_records))])

    n_splits = max(2, min(N_SPLITS, len(train_records)))
    gkf = GroupKFold(n_splits=n_splits)

    C, gamma = C_GRID[0], GAMMA_GRID[0]
    seed = SEED_BASE + fold_idx

    # OOF scores for calibration. We need OOF scores over the FULL train rows,
    # so cross_val_predict runs on the full (unsubsampled) X to keep every train
    # row scored; the per-fold FIT inside it is on subsampled negatives via a
    # custom split is not exposed by cross_val_predict, so we replicate manually.
    oof_scores = np.empty(len(y_train_all), dtype=float)
    for tr_idx, val_idx in gkf.split(X_train_all, y_train_all, groups):
        Xs, ys = subsample_negatives(X_train_all[tr_idx], y_train_all[tr_idx],
                                     NEG_PER_POS, seed)
        pipe = make_pipeline(C, gamma, k_features)
        pipe.fit(Xs, ys)
        oof_scores[val_idx] = pipe.decision_function(X_train_all[val_idx])

    th_high = choose_th_high(oof_scores, y_train_all)

    train_recs = [
        build_record_prediction(r, oof_scores[bounds[i]:bounds[i + 1]])
        for i, r in enumerate(train_records)
    ]
    best_k, best_th_low = calibrate_k_and_threshold(
        train_recs, th_high=th_high,
        target_fp_per_day=TARGET_FP_PER_DAY, k_grid=K_GRID
    )

    # Final model on subsampled negatives of the full training pool
    Xs, ys = subsample_negatives(X_train_all, y_train_all, NEG_PER_POS, seed)
    pipe = make_pipeline(C, gamma, k_features)
    pipe.fit(Xs, ys)
    scores = pipe.decision_function(X_test)
    y_pred = (scores >= 0).astype(int)

    acc = accuracy_score(y_test, y_pred)
    auc = safe_auc(y_test, scores)
    auprc = safe_auprc(y_test, scores)
    precision = precision_score(y_test, y_pred, zero_division=0)
    sensitivity = recall_score(y_test, y_pred, zero_division=0)
    tn, fp, fn, tp = confusion_matrix(y_test, y_pred, labels=[0, 1]).ravel()
    specificity = tn / (tn + fp) if (tn + fp) > 0 else np.nan

    seg = {"acc": acc, "auc": auc, "auprc": auprc,
           "sensitivity": sensitivity, "specificity": specificity, "precision": precision}

    rec = build_record_prediction(test_record, scores)
    out = score_one_file(rec, th_low=best_th_low, th_high=th_high, k_consecutive=best_k)

    th_high_disp = "inf" if not np.isfinite(th_high) else f"{th_high:.3f}"
    log = (f"[{test_record}] kfeat={k_features} k={best_k} "
           f"th_low={best_th_low:.3f} th_high={th_high_disp} | "
           f"AUPRC={auprc:.3f} | TP/FP={out['tp']}/{out['fp']} "
           f"sens={out['sensitivity']} FP/day={out['fp_per_day']} lat={out['median_latency']}")
    return seg, out, log

# =====================================================
# DRIVER
# =====================================================

# S2: choose k_features once, up front
K_FEATURES = choose_k_features_once(records)
print(f"Chosen k_features (once, full pool): {K_FEATURES}")

# S1: parallelize the outer LORO loop
rec_ids = list(records)
parallel_out = Parallel(n_jobs=N_OUTER_JOBS, backend="loky")(
    delayed(run_one_fold)(i, test_record, records, K_FEATURES)
    for i, test_record in enumerate(rec_ids)
)

results = [seg for seg, _out, _log in parallel_out]
event_results = [out for _seg, out, _log in parallel_out]
for _seg, _out, log in parallel_out:
    print(log)

# =====================================================
# FINAL SUMMARY  (unchanged)
# =====================================================

print("\n================ SEGMENT SUMMARY ================")
print("Mean ACC  :", np.nanmean([r["acc"] for r in results]))
print("Mean AUC  :", np.nanmean([r["auc"] for r in results]))
print("Mean AUPRC:", np.nanmean([r["auprc"] for r in results]))
print("Mean Sens :", np.nanmean([r["sensitivity"] for r in results]))
print("Mean Spec :", np.nanmean([r["specificity"] for r in results]))
print("Mean Prec :", np.nanmean([r["precision"] for r in results]))

print("\n================ EVENT SUMMARY (SzCORE, per-subject mean) ================")
by_pt = defaultdict(lambda: {"tp": 0, "fp": 0, "n_ref": 0, "hours": 0.0})
for r in event_results:
    p = by_pt[r["patient"]]
    p["tp"] += r["tp"]
    p["fp"] += r["fp"]
    p["n_ref"] += r["n_ref"]
    p["hours"] += r["duration_hours"]

sens_list, prec_list, f1_list, fppd_list = [], [], [], []
for pid, p in sorted(by_pt.items()):
    sens = p["tp"] / p["n_ref"] if p["n_ref"] else np.nan
    prec = p["tp"] / (p["tp"] + p["fp"]) if (p["tp"] + p["fp"]) else np.nan
    f1 = (2 * prec * sens / (prec + sens)
          if (prec is not np.nan and sens is not np.nan and (prec + sens)) else np.nan)
    fppd = p["fp"] / (p["hours"] / 24) if p["hours"] else np.nan
    sens_list.append(sens); prec_list.append(prec); f1_list.append(f1); fppd_list.append(fppd)
    print(f"  {pid}: sens={sens:.3f} prec={prec:.3f} f1={f1:.3f} fp/day={fppd:.3f}")

print("Mean Sensitivity:", np.nanmean(sens_list))
print("Mean Precision  :", np.nanmean(prec_list))
print("Mean F1         :", np.nanmean(f1_list))
print("Mean FP/day     :", np.nanmean(fppd_list))
print("Median latency  :", np.nanmedian([r["median_latency"] for r in event_results]))
```

```python
"""
Leave-one-record-out (LORO) seizure-detection pipeline — SzCORE-aligned scoring,
leakage-free DUAL-THRESHOLD detector — SPEED-OPTIMIZED.

Method is UNCHANGED. Only the execution is faster:
  S1. OUTER LORO LOOP IS PARALLELIZED across test records (joblib). Each fold
      is independent. Inner SVM/BLAS threads are pinned to 1 to avoid
      oversubscription; outer parallelism does the work.
  S2. k_features (SelectKBest) IS CHOSEN ONCE on the full pool, not swept per
      fold. The feature ranking is stable across the near-identical LORO
      training pools, so per-fold re-selection only added ~KFEAT x n_splits
      redundant SVM fits. Selection is still done on training-style data
      (GroupKFold AUC), never on any single held-out test record.
  S3. NEGATIVE SUBSAMPLING in the training fold: interictal windows are
      randomly subsampled to NEG_PER_POS x the positive count, with a fixed
      per-fold seed (deterministic, reproducible). class_weight='balanced'
      is retained. This shrinks the RBF kernel from O(n^2) on a ~99%-negative
      matrix. Positives are never dropped; the TEST record is never subsampled.

Scoring (timescoring/SzCORE), the dual-threshold detector, th_high precision
gate, th_low+k calibration, latency, and per-subject aggregation are byte-for-
byte identical to the previous version.

NOTE on the leakage-comparison experiment: subsampling changes the training set,
so the SAME NEG_PER_POS and SEED_BASE must be used across all three conditions,
or the comparison gains a confound.
"""

import os
# Pin math-library threads BEFORE numpy import so outer joblib parallelism
# doesn't oversubscribe cores (inner work stays single-threaded).
os.environ.setdefault("OMP_NUM_THREADS", "1")
os.environ.setdefault("OPENBLAS_NUM_THREADS", "1")
os.environ.setdefault("MKL_NUM_THREADS", "1")
os.environ.setdefault("NUMEXPR_NUM_THREADS", "1")

from dataclasses import dataclass
import re
from collections import defaultdict

import numpy as np
import pandas as pd

from sklearn.svm import SVC
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.model_selection import GroupKFold, cross_val_predict
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    average_precision_score,
    recall_score,
    precision_score,
    confusion_matrix,
)

from timescoring import scoring
from timescoring.annotations import Annotation

from joblib import Parallel, delayed

# =====================================================
# PARAMETERS
# =====================================================

WIN_SEC = 10.0
STEP_SEC = 2.5

KFEAT_GRID = [20, 50, 100, 200]   # SelectKBest sweep (chosen ONCE, see S2)

C_GRID = [0.1,1,10]
GAMMA_GRID = ["scale"]

# Event-level operating point (chosen on TRAIN, applied to TEST)
TARGET_FP_PER_DAY = 1.0
K_GRID = [2, 4, 6]
THRESHOLD_GRID = np.linspace(-2.0, 2.0, 41)

TH_HIGH_PRECISION = 0.60          # confidence gate. 0.90 was unreachable on real OOF scores
                                  # (gate returned inf every fold -> dual-threshold did nothing).
                                  # 0.60 is reachable; still a-priori, training-derived, not test-tuned.

TOL_PRE = 30.0
TOL_POST = 60.0

SZCORE_FS = 1
N_SPLITS = 5

# Speed knobs
N_OUTER_JOBS = 8                  # parallel LORO folds (set to your core count)
NEG_PER_POS = 10000               # effectively OFF: keep all negatives (no boundary shift).
                                  # Lower (e.g. 40) only if runtime forces it, and re-verify
                                  # sensitivity matches this baseline before trusting results.
SEED_BASE = 12345                 # base RNG seed; per-fold seed = SEED_BASE + fold_idx

ann = pd.read_csv("seizure_annotations.csv")

# =====================================================
# STRUCTURE
# =====================================================

@dataclass
class RecordPrediction:
    record_id: str
    window_starts: np.ndarray
    window_len: float
    scores: np.ndarray
    duration_sec: float
    true_seizures: list

# =====================================================
# HELPERS
# =====================================================

def _runs_of_true(mask):
    runs = []
    i = 0
    while i < len(mask):
        if mask[i]:
            j = i
            while j + 1 < len(mask) and mask[j + 1]:
                j += 1
            runs.append((i, j))
            i = j + 1
        else:
            i += 1
    return runs


def safe_auc(y_true, scores):
    return roc_auc_score(y_true, scores) if len(np.unique(y_true)) > 1 else np.nan


def safe_auprc(y_true, scores):
    return average_precision_score(y_true, scores) if len(np.unique(y_true)) > 1 else np.nan


def patient_of(record_id):
    m = re.match(r"(chb\d+)", record_id)
    return m.group(1) if m else record_id


def build_record_prediction(record_id, scores):
    n = len(scores)
    window_starts = np.arange(n) * STEP_SEC
    duration_sec = (window_starts[-1] + WIN_SEC) if n else 0.0
    return RecordPrediction(
        record_id=record_id,
        window_starts=window_starts,
        window_len=WIN_SEC,
        scores=np.asarray(scores),
        duration_sec=duration_sec,
        true_seizures=get_seizures_for_record(ann, record_id),
    )

# =====================================================
# DUAL-THRESHOLD DETECTOR -> 1 Hz ALARM MASK  (unchanged)
# =====================================================

def _alarm_runs(rec, th_low, th_high, k_consecutive):
    weak = rec.scores >= th_low
    strong = rec.scores >= th_high
    alarms = []
    for s, e in _runs_of_true(weak):
        run_len = e - s + 1
        strong_idx = np.where(strong[s:e + 1])[0]
        has_strong = strong_idx.size > 0
        if not (run_len >= k_consecutive or has_strong):
            continue
        persist_decision_idx = s + (k_consecutive - 1) if run_len >= k_consecutive else None
        strong_decision_idx = s + strong_idx[0] if has_strong else None
        candidates = [c for c in (persist_decision_idx, strong_decision_idx) if c is not None]
        decision_idx = min(candidates)
        start_sec = rec.window_starts[s]
        decision_sec = rec.window_starts[decision_idx] + rec.window_len
        end_sec = rec.window_starts[e] + rec.window_len
        alarms.append((start_sec, decision_sec, end_sec))
    return alarms


def alarm_mask(rec, th_low, th_high, k_consecutive, fs=SZCORE_FS):
    n = int(np.ceil(rec.duration_sec)) * fs
    mask = np.zeros(n, dtype=int)
    for start_sec, _decision_sec, end_sec in _alarm_runs(rec, th_low, th_high, k_consecutive):
        a = int(round(start_sec * fs))
        b = int(round(end_sec * fs))
        mask[a:min(b, n)] = 1
    return mask


def reference_mask(rec, fs=SZCORE_FS):
    n = int(np.ceil(rec.duration_sec)) * fs
    mask = np.zeros(n, dtype=int)
    for sz_start, sz_end in rec.true_seizures:
        mask[int(round(sz_start * fs)):min(int(round(sz_end * fs)), n)] = 1
    return mask

# =====================================================
# SECONDARY: detection latency  (unchanged)
# =====================================================

def first_alarm_latency(rec, th_low, th_high, k_consecutive):
    alarms = sorted(_alarm_runs(rec, th_low, th_high, k_consecutive))
    latencies = []
    for sz_start, sz_end in rec.true_seizures:
        lo, hi = sz_start - TOL_PRE, sz_end + TOL_POST
        best = None
        for start_sec, decision_sec, end_sec in alarms:
            if start_sec < hi and end_sec > lo:
                best = max(0.0, decision_sec - sz_start)
                break
        if best is not None:
            latencies.append(best)
    return np.median(latencies) if latencies else np.nan

# =====================================================
# EVENT SCORING (SzCORE via timescoring)  (unchanged)
# =====================================================

def score_one_file(rec, th_low, th_high, k_consecutive, fs=SZCORE_FS):
    ref = Annotation(reference_mask(rec, fs), fs)
    hyp = Annotation(alarm_mask(rec, th_low, th_high, k_consecutive, fs), fs)
    s = scoring.EventScoring(ref, hyp)
    return {
        "record_id": rec.record_id,
        "patient": patient_of(rec.record_id),
        "tp": s.tp,
        "fp": s.fp,
        "n_ref": s.refTrue,
        "duration_hours": rec.duration_sec / 3600,
        "sensitivity": s.sensitivity,
        "precision": s.precision,
        "f1": s.f1,
        "fp_per_day": s.fpRate,
        "median_latency": first_alarm_latency(rec, th_low, th_high, k_consecutive),
    }

# =====================================================
# th_high precision gate  (unchanged)
# =====================================================

def choose_th_high(oof_scores, y_train, precision_target=TH_HIGH_PRECISION,
                   candidate_thresholds=THRESHOLD_GRID):
    y = np.asarray(y_train)
    for th in np.sort(candidate_thresholds):
        pred = oof_scores >= th
        tp = np.sum(pred & (y == 1))
        fp = np.sum(pred & (y == 0))
        if (tp + fp) == 0:
            continue
        prec = tp / (tp + fp)
        if prec >= precision_target:
            return float(th)
    return np.inf

# =====================================================
# (k, th_low) calibration  (unchanged)
# =====================================================

def calibrate_k_and_threshold(train_recs, th_high, target_fp_per_day, k_grid,
                              candidate_thresholds=THRESHOLD_GRID):
    best = None
    for k in k_grid:
        for th in np.sort(candidate_thresholds):
            tp = fp = n_ref = 0
            total_hours = 0.0
            for rec in train_recs:
                o = score_one_file(rec, th_low=th, th_high=th_high, k_consecutive=k)
                tp += o["tp"]
                fp += o["fp"]
                n_ref += o["n_ref"]
                total_hours += o["duration_hours"]
            fp_per_day = fp / (total_hours / 24) if total_hours else np.inf
            if fp_per_day <= target_fp_per_day:
                sens = tp / n_ref if n_ref else 0.0
                if best is None or sens > best[0]:
                    best = (sens, k, th)
                break
    if best is None:
        return k_grid[0], float(np.max(candidate_thresholds))
    return best[1], best[2]

# =====================================================
# PIPELINE + NEGATIVE SUBSAMPLING
# =====================================================

def make_pipeline(C, gamma, k_features):
    return Pipeline([
        ("select", SelectKBest(f_classif, k=k_features)),
        ("scale", StandardScaler()),
        ("svc", SVC(C=C, gamma=gamma, kernel="rbf",
                    class_weight="balanced", cache_size=4000)),
    ])


def subsample_negatives(X, y, neg_per_pos, seed):
    """Keep all positives; randomly keep neg_per_pos * n_pos negatives.
    Deterministic given seed. If there aren't that many negatives, keep all."""
    y = np.asarray(y)
    pos_idx = np.where(y == 1)[0]
    neg_idx = np.where(y == 0)[0]
    # If a fold has no positives, subsampling would empty the negatives too.
    # Keep the data unchanged in that degenerate case (the fold is single-class
    # anyway and contributes nan to AUC via the safe_auc guard).
    if len(pos_idx) == 0:
        return X, y
    n_keep = min(len(neg_idx), neg_per_pos * len(pos_idx))
    rng = np.random.default_rng(seed)
    keep_neg = rng.choice(neg_idx, size=n_keep, replace=False) if n_keep else np.array([], int)
    keep = np.sort(np.concatenate([pos_idx, keep_neg]))
    return X[keep], y[keep]

# =====================================================
# k_features CHOSEN ONCE (S2)
# =====================================================

def choose_k_features_once(records_dict):
    """Pick k_features on the full pooled dataset via GroupKFold AUC, once,
    before the LORO loop. Uses every record as its own group; this is a
    training-style selection (no single record is treated as the held-out
    test here)."""
    rec_ids = list(records_dict)
    X = np.vstack([records_dict[r][0] for r in rec_ids])
    y = np.concatenate([records_dict[r][1] for r in rec_ids])
    groups = np.concatenate([
        np.full(len(records_dict[r][1]), i) for i, r in enumerate(rec_ids)
    ])
    n_splits = max(2, min(N_SPLITS, len(rec_ids)))
    gkf = GroupKFold(n_splits=n_splits)

    best_auc, best_kfeat = -np.inf, KFEAT_GRID[0]
    for k_features in KFEAT_GRID:
        fold_aucs = []
        for tr_idx, val_idx in gkf.split(X, y, groups):
            Xs, ys = subsample_negatives(X[tr_idx], y[tr_idx], NEG_PER_POS, SEED_BASE)
            pipe = make_pipeline(C_GRID[0], GAMMA_GRID[0], k_features)
            pipe.fit(Xs, ys)
            sc = pipe.decision_function(X[val_idx])
            fold_aucs.append(safe_auc(y[val_idx], sc))
        mean_auc = np.nanmean(fold_aucs)
        if mean_auc > best_auc:
            best_auc, best_kfeat = mean_auc, k_features
    return best_kfeat

# =====================================================
# ONE LORO FOLD (runs in a worker)
# =====================================================

def run_one_fold(fold_idx, test_record, records_dict, k_features):
    X_test, y_test = records_dict[test_record]
    train_records = [r for r in records_dict if r != test_record]

    X_train_all = np.vstack([records_dict[r][0] for r in train_records])
    y_train_all = np.concatenate([records_dict[r][1] for r in train_records])

    lengths = [len(records_dict[r][1]) for r in train_records]
    bounds = np.cumsum([0] + lengths)
    groups = np.concatenate([np.full(lengths[i], i) for i in range(len(train_records))])

    n_splits = max(2, min(N_SPLITS, len(train_records)))
    gkf = GroupKFold(n_splits=n_splits)

    C, gamma = C_GRID[0], GAMMA_GRID[0]
    seed = SEED_BASE + fold_idx

    # OOF scores for calibration. We need OOF scores over the FULL train rows,
    # so cross_val_predict runs on the full (unsubsampled) X to keep every train
    # row scored; the per-fold FIT inside it is on subsampled negatives via a
    # custom split is not exposed by cross_val_predict, so we replicate manually.
    oof_scores = np.empty(len(y_train_all), dtype=float)
    for tr_idx, val_idx in gkf.split(X_train_all, y_train_all, groups):
        Xs, ys = subsample_negatives(X_train_all[tr_idx], y_train_all[tr_idx],
                                     NEG_PER_POS, seed)
        pipe = make_pipeline(C, gamma, k_features)
        pipe.fit(Xs, ys)
        oof_scores[val_idx] = pipe.decision_function(X_train_all[val_idx])

    th_high = choose_th_high(oof_scores, y_train_all)

    train_recs = [
        build_record_prediction(r, oof_scores[bounds[i]:bounds[i + 1]])
        for i, r in enumerate(train_records)
    ]
    best_k, best_th_low = calibrate_k_and_threshold(
        train_recs, th_high=th_high,
        target_fp_per_day=TARGET_FP_PER_DAY, k_grid=K_GRID
    )

    # Final model on subsampled negatives of the full training pool
    Xs, ys = subsample_negatives(X_train_all, y_train_all, NEG_PER_POS, seed)
    pipe = make_pipeline(C, gamma, k_features)
    pipe.fit(Xs, ys)
    scores = pipe.decision_function(X_test)
    y_pred = (scores >= 0).astype(int)

    acc = accuracy_score(y_test, y_pred)
    auc = safe_auc(y_test, scores)
    auprc = safe_auprc(y_test, scores)
    precision = precision_score(y_test, y_pred, zero_division=0)
    sensitivity = recall_score(y_test, y_pred, zero_division=0)
    tn, fp, fn, tp = confusion_matrix(y_test, y_pred, labels=[0, 1]).ravel()
    specificity = tn / (tn + fp) if (tn + fp) > 0 else np.nan

    seg = {"acc": acc, "auc": auc, "auprc": auprc,
           "sensitivity": sensitivity, "specificity": specificity, "precision": precision}

    rec = build_record_prediction(test_record, scores)
    out = score_one_file(rec, th_low=best_th_low, th_high=th_high, k_consecutive=best_k)

    th_high_disp = "inf" if not np.isfinite(th_high) else f"{th_high:.3f}"
    log = (f"[{test_record}] kfeat={k_features} k={best_k} "
           f"th_low={best_th_low:.3f} th_high={th_high_disp} | "
           f"AUPRC={auprc:.3f} | TP/FP={out['tp']}/{out['fp']} "
           f"sens={out['sensitivity']} FP/day={out['fp_per_day']} lat={out['median_latency']}")
    return seg, out, log

# =====================================================
# DRIVER
# =====================================================

# S2: choose k_features once, up front
K_FEATURES = choose_k_features_once(records)
print(f"Chosen k_features (once, full pool): {K_FEATURES}")

# S1: parallelize the outer LORO loop
rec_ids = list(records)
parallel_out = Parallel(n_jobs=N_OUTER_JOBS, backend="loky")(
    delayed(run_one_fold)(i, test_record, records, K_FEATURES)
    for i, test_record in enumerate(rec_ids)
)

results = [seg for seg, _out, _log in parallel_out]
event_results = [out for _seg, out, _log in parallel_out]
for _seg, _out, log in parallel_out:
    print(log)

# =====================================================
# FINAL SUMMARY  (unchanged)
# =====================================================

print("\n================ SEGMENT SUMMARY ================")
print("Mean ACC  :", np.nanmean([r["acc"] for r in results]))
print("Mean AUC  :", np.nanmean([r["auc"] for r in results]))
print("Mean AUPRC:", np.nanmean([r["auprc"] for r in results]))
print("Mean Sens :", np.nanmean([r["sensitivity"] for r in results]))
print("Mean Spec :", np.nanmean([r["specificity"] for r in results]))
print("Mean Prec :", np.nanmean([r["precision"] for r in results]))

print("\n================ EVENT SUMMARY (SzCORE, per-subject mean) ================")
by_pt = defaultdict(lambda: {"tp": 0, "fp": 0, "n_ref": 0, "hours": 0.0})
for r in event_results:
    p = by_pt[r["patient"]]
    p["tp"] += r["tp"]
    p["fp"] += r["fp"]
    p["n_ref"] += r["n_ref"]
    p["hours"] += r["duration_hours"]

sens_list, prec_list, f1_list, fppd_list = [], [], [], []
for pid, p in sorted(by_pt.items()):
    sens = p["tp"] / p["n_ref"] if p["n_ref"] else np.nan
    prec = p["tp"] / (p["tp"] + p["fp"]) if (p["tp"] + p["fp"]) else np.nan
    f1 = (2 * prec * sens / (prec + sens)
          if (prec is not np.nan and sens is not np.nan and (prec + sens)) else np.nan)
    fppd = p["fp"] / (p["hours"] / 24) if p["hours"] else np.nan
    sens_list.append(sens); prec_list.append(prec); f1_list.append(f1); fppd_list.append(fppd)
    print(f"  {pid}: sens={sens:.3f} prec={prec:.3f} f1={f1:.3f} fp/day={fppd:.3f}")

print("Mean Sensitivity:", np.nanmean(sens_list))
print("Mean Precision  :", np.nanmean(prec_list))
print("Mean F1         :", np.nanmean(f1_list))
print("Mean FP/day     :", np.nanmean(fppd_list))
print("Median latency  :", np.nanmedian([r["median_latency"] for r in event_results]))
```
